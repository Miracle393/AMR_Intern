PREVIEW THE PATIENT THAT ARE SINGLE RESISTANT FROM RAW DATASET

```python
import pandas as pd
```

```python
amr = pd.read_excel("AMR DATA 7.7.2026.xlsx")
```

```python
amr.columns
```

```python
# Store the antibiotic columns
antibiotics = [
    'Amikacin',
    'Amoxicillin/Clavulanic acid',
    'Ampicillin',
    'AZM',
    'Cefazolin',
    'Cefepime',
    'Cefixime',
    'Cefotaxime',
    'Cefoxitin',
    'Ceftazidime',
    'Ceftriaxone',
    'Chloramphenicol',
    'Ciprofloxacin',
    'Clindamycin',
    'Erythromycin',
    'GEH',
    'Gentamicin',
    'Imipenem',
    'Meropenem',
    'Nitrofurantoin',
    'PEN',
    'Piperacillin/Tazobactam',
    'Tetracycline',
    'Trimethoprim/Sulfamethoxazole',
    'VAN'
]
```

```python
r_count = (amr[antibiotics] == 'Resistant').sum(axis=1)
```

```python
r_count
```

```python
single_resistant = amr[r_count == 1].copy()
```

```python
single_resistant.shape
```

```python
(r_count == 1).sum()
```

Which antibiotic is the one they're resistant to?"

```python
#new column to specific name of resisted antibiotic
single_resistant['Resistant Antibiotic'] = (
    single_resistant[antibiotics]
    .eq('Resistant')
    .idxmax(axis=1)
)
```

```python
single_resistant['Resistant Antibiotic'].value_counts()
```

```python
#save it
#single_resistant.to_csv( "single_resistant_patients.csv", index=False)
```

```python
single_resistant['Resistant Antibiotic'].value_counts().sum()
```

```python
#Standardize
mapping = {
    'R': 'Resistant',
    'Resistant': 'Resistant',

    'S': 'Susceptible',
    'Susceptible': 'Susceptible',

    'I': 'Intermediate',
    'Intermediate': 'Intermediate',

    1: 'Resistant',
    0: 'Susceptible',
    
    '1': 'Resistant',
    '0': 'Susceptible'
    
}
```

```python
#Aply
amr[antibiotics] = amr[antibiotics].replace(mapping)
```
