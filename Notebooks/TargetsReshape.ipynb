```python
target_mapping = {
    '1': 'Resistant',
    'R': 'Resistant',
    'r': 'Resistant',
    'Resistant': 'Resistant',
    'resistant': 'Resistant',

    '0': 'Susceptible',
    'S': 'Susceptible',
    's': 'Susceptible',
    'Susceptible': 'Susceptible',
    'susceptible': 'Susceptible',

    'I': 'Intermediate',
    'i': 'Intermediate',
    'Intermediate': 'Intermediate',
    'intermediate': 'Intermediate'
}
```

```python
antibiotics = [
    'Amikacin',
    'Amoxicillin/Clavulanic acid',
    'Ampicillin',
    'AZM', 'Cefazolin',
    'Cefepime',
    'Cefixime',
    'Cefotaxime',
    'Cefoxitin',
    'Ceftazidime',
    'Ceftriaxone', 
    'Chloramphenicol',
    'Ciprofloxacin',
    'Clindamycin',
    'Erythromycin',
    'GEH',
    'Gentamicin',
    'Imipenem',
    'Meropenem',
    'Nitrofurantoin',
    'PEN',
    'Piperacillin/Tazobactam',
    'Tetracycline',
    'Trimethoprim/Sulfamethoxazole',
    'VAN'
]
```

```python
for col in antibiotics:
   ts[col] = ts[col].map(target_mapping)
```
