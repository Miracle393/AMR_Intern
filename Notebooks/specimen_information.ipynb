THERE IS RESERCHER DECISION ON 'SPECIMEN CATEGORY' INPUT, WHERE HERE I GOT TO CREATE NEW COLUMN FOR THIS ACCORDING TO [ URINE = UTI [Urinary Tract Infection] , BLOOD = BSI [Blood Stream Infection ], PUS = SSTI [Skin and Soft Tissue Infection]]

```python
sp_clean['Specimen type'].unique()
```

```python
#create dictionary first
specimen_category = {
    'urine': 'Urinary Tract Infection',
    'Blood': 'Blood Stream Infection',
    'Pus': 'Skin and Soft Tissue Infection'
}
```

```python
#create new column
sp_clean['Specimen Category'] = (
    sp_clean['Specimen type']
        .map(specimen_category)
)
```

```python
sp_clean['Specimen Category'].unique()
#sp_clean['Specimen Category']
```

```python
sp_clean.columns
```

```python
#create dictionary for specimen code
category_code = {
    'urine':'UTI',
    'Blood':'BSI',
    'Pus':'SSTI'
}
```

```python
sp_clean['Specimen Category Code'] = (
    sp_clean['Specimen type']
        .map(category_code)
)
```

```python
sp_clean['Specimen Category Code'].unique()
```

Other important thing to deal with in specimen information is Specimen date, though is not written in implementation docx file but it carry weight to model to determin the season of when the specimen was collected.

So here is to standardize the date on how to write them in this dataset, though thwy came with time in raw data but the time is same all which is 0000hrs but we got to remove the time and remain with dates only and standardize on how to write those dates in their column.

```python
#contniue with recent cleaned csv file
sp_sd = pd.read_csv('specimen_information_clean_3.csv')
```

- Standardize Specimen date
- Original data contained mixed date formats (DD/MM/YYYY and YYYY-M-D)
- Converted to datetime, then formatted as DD/MM/YYYY according to the research team's standard.

```python
#Convert the data type in this column to be same standrd but in mixed formart
sp_sd['Specimen date'] = pd.to_datetime(
    sp_sd['Specimen date'], dayfirst= True, format= 'mixed'
    )
```

```python
#verify
sp_sd['Specimen date'].dtype
```

```python
sp_sd['Specimen date'].unique()
```

```python
sp_sd['Specimen date'].sample(10, random_state=42)
```

```python
#change date to team standard and remove time
sp_sd['Specimen date'] = sp_sd['Specimen date'].dt.strftime('%d/%m/%Y')
```
