```python
import pandas as pd
import os
```

```python
ds = pd.read_csv(r"../Cleaned amr data\Targets_stand.csv")
```

```python
ds['Sentinel Site'].unique()
```

```python
district_mapping = {
    "Bugando Medical Centre": "Nyamagana",
    "Muhimbili National Hospital": "Ilala",
    "Kilimanjaro Christian Medical Centre": "Moshi",
    "Benjamin Mkapa Hospital": "Dodoma",
    "Morogoro Regional Referral Hospital": "Morogoro",
    "Temeke Regional Referral Hospital": "Temeke",
    "Mbeya Zonal Referral Hospital": "Mbeya",
    "Kigoma Regional Hospital": "Kigoma",
    "Ligula Regional Referral Hospital": "Mtwara"
}
```

```python
ds["District"] = ds["Sentinel Site"].map(district_mapping)
```

```python
ds["District"].value_counts(dropna=False)
```

```python
district = ds.pop("District")

position = ds.columns.get_loc("Region") + 1

ds.insert(position, "District", district)
```
