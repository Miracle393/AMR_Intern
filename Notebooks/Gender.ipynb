GOT TO CLEAN AND RESHAPE THE SEX COLUMN HAVING f,m,NaN (3 rows)

```python
import pandas as pd
import os
```

```python
pm_sex = pd.read_csv(r"../Cleaned amr data\Age_Clean_v2.csv")
```

```python
#inspect the data
pm_sex['Sex'].value_counts(dropna=False)
```

```python
#Standardize the gender
pm_sex['Sex'] = pm_sex['Sex'].replace({
    'f': 'F',
    'm': 'M'
})
```

```python
#Verify the changes
pm_sex['Sex'].value_counts(dropna=False)
```

I need to fill the three NaN in the Sex Column, where two by names clue I got to know are male exept one

```python
#look for columns
pm_sex.columns
```

```python
#cut the names part
pm_sex.loc[
    pm_sex['Sex'].isna(),
    [
        'Sex',
        'First name',
        'Last name',
        'Identification number'
    ]
]
```

```python
#fill the 2 NaN
pm_sex.loc[3068, 'Sex'] = 'M'
pm_sex.loc[3909, 'Sex'] = 'M'
```

```python
#Verify the changes
pm_sex.loc[
    pm_sex['Sex'].isna(),
    [
        'Sex',
        'First name',
        'Last name',
        'Identification number'
    ]
]
```

```python
#save the cleaned data
#pm_sex.to_csv(r"../Cleaned amr data\Sex_with_na.csv", index=False)
```

I leave one row with NAN, use Mode Imputation to fill the 1 Nan

```python
#mode
pm_sex['Sex'].mode()
```

```python
#fill with mode
pm_sex['Sex'] = pm_sex['Sex'].fillna(
    pm_sex['Sex'].mode()[0]
)
```

```python
#verify the changes
pm_sex['Sex'].value_counts(dropna=False)
```
