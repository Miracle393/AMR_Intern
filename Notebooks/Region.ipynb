```python
import pandas as pd
import os
```

```python
rg = pd.read_csv(r"../Cleaned amr data\SentinelSites.csv")
```

```python
#mapping region
regions = {
    'BMC': 'MWANZA',
    'MNH': 'DAR ES SALAAM',
    'KCMC': 'KILIMANJARO',
    'BMHL': 'DODOMA',
    'MoroRRH': 'MOROGORO',
    'TMK': 'DAR ES SALAAM',
    'MZRH': 'MBEYA',
    'KGM': 'KIGOMA',
    'Ligula RRH': 'MTWARA',
}
```

```python
#create region column
rg['Region'] = rg['Hospital Name'].map(regions)
```

```python
rg['Region'].value_counts(dropna = False)
```

```python
#move the column to the front
column = rg.pop('Region')

position = rg.columns.get_loc('Origin')

rg.insert(position, 'Region', column)
```

```python
#save the dataframe to a new CSV file
rg.to_csv(r"../Cleaned amr data\Region.csv", index=False)
```
