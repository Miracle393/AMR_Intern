```python
import pandas as pd

sp_date = pd.read_csv("specimen_information_clean_3.csv")
```

```python
#inspect
sp_date['Specimen date'].sample(10, random_state=42)
```

```python
#crate masks
slash_mask = sp_date['Specimen date'].str.contains('/', na=False)

dash_mask = sp_date['Specimen date'].str.contains('-', na=False)
```

```python
#check
print("Slash format:", slash_mask.sum())
print("Dash format :", dash_mask.sum())
```

```python
"""
#convert with /
sp_date.loc[slash_mask, 'Specimen date'] = pd.to_datetime(
    sp_date.loc[slash_mask, 'Specimen date'],
    dayfirst=True
)
"""
```

```python
sp_date['Specimen date'] = sp_date['Specimen date'].str.strip()
```

```python
sp_date['Specimen date'] = pd.to_datetime(
    sp_date['Specimen date'],
    format='mixed',
    dayfirst=True
)
```

```python
sp_date['Specimen date'].head(10)
```

```python
sp_date.loc[
    sp_date['Specimen date'].dt.month == 2,
    'Specimen date'
].head(20)
```

```python
sp_date.loc[
    sp_date['Specimen date'].dt.month == 4,
    'Specimen date'
].head(20)
```

```python
sp_date["Specimen date"] = (
    sp_date["Specimen date"]
        .dt.strftime("%d/%m/%Y")
)
```
