```python
# Convert mixed-format strings to datetime
cm['Specimen date'] = pd.to_datetime(
    cm['Specimen date'],
    format='mixed',
    dayfirst=True
)

# Create Collection Year
cm['Collection Year'] = cm['Specimen date'].dt.year

# Create Collection Month
cm['Collection Month'] = cm['Specimen date'].dt.month_name()
```

```python
cm['Collection Year'].value_counts(dropna = False)
```

```python
cm['Collection Month'].value_counts(dropna = False)
```

```python
cm['Collection Quarter'] = (
    'Q' + cm['Specimen date'].dt.quarter.astype(str)
)
```

```python
cm['Collection Quarter'].value_counts(dropna = False)
```

```python
cm['Collection Season'] = (
    cm['Specimen date']
    .dt.month
    .map({
        1: 'Rainy Season',
        2: 'Rainy Season',
        3: 'Rainy Season',
        4: 'Rainy Season',
        5: 'Rainy Season',
        6: 'Dry Season',
        7: 'Dry Season',
        8: 'Dry Season',
        9: 'Dry Season',
        10: 'Dry Season',
        11: 'Rainy Season',
        12: 'Rainy Season'
    })
)
```

```python
cm['Collection Season'].value_counts(dropna = False)
```

```python
cm.columns
```

```python
ltp = cm.columns.get_loc('Location type')

cy= cm.pop('Collection Year')
cmm = cm.pop('Collection Month')
cq = cm.pop('Collection Quarter')
cs = cm.pop('Collection Season')

cm.insert(ltp + 1, 'Collection Year', cy)
cm.insert(ltp + 2, 'Collection Month', cmm)
cm.insert(ltp + 3, 'Collection Quarter', cq)
cm.insert(ltp + 4, 'Collection Season', cs)
```
