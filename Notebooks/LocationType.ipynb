clean the in_patient and out_patient

```python
import pandas as pd
import os
```

```python
pm_oi = pd.read_csv(r"../Cleaned amr data\Sex_Clean.csv")
```

```python
#inpsect the data
pm_oi['Location type'].value_counts(dropna=False)
```

```python
pm_oi.loc[
    pm_oi['Location type'] == 'mix',
    [
        'Identification number',
        'Hospital Name',
        'Ward',
        'Department',
        'Specimen type',
        'Location type'
    ]
]
```

since the ward is clinic means is mostly to be out_patient

```python
#Standardize the 'mix' 
pm_oi['Location type'] = pm_oi['Location type'].replace('mix', 'out_patient') 
```

```python
#verify the changes
pm_oi.loc[
    pm_oi['Location type'] == 'mix',
    [
        'Identification number',
        'Hospital Name',
        'Ward',
        'Department',
        'Specimen type',
        'Location type'
    ]
]
```

```python
pm_oi['Location type'].value_counts(dropna=False)
```

fill nan with mode since no other clue from other columns

```python
pm_oi['Location type'].mode()
```

```python
#fill nan with mode
pm_oi['Location type'] = pm_oi['Location type'].fillna(
    pm_oi['Location type'].mode()[0])
```

```python
#verify the changes
pm_oi['Location type'].value_counts(dropna=False)
```
