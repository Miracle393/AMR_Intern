```python
import pandas as pd
import os
```

```python
hn = pd.read_csv(r"../Cleaned amr data\Department_w_nan.csv")
```

```python
hn['Hospital Name'].value_counts(dropna=False)
```

```python
#map the long term
hospital_map = {
    'BMC': 'Bugando Medical Centre',
    'MNH': 'Muhimbili National Hospital',
    'KCMC': 'Kilimanjaro Christian Medical Centre',
    'BMHL': 'Benjamin Mkapa Hospital',
    'MoroRRH': 'Morogoro Regional Referral Hospital',
    'TMK': 'Temeke Regional Referral Hospital',
    'MZRH': 'Mbeya Zonal Referral Hospital',
    'KGM': 'Kigoma Regional Hospital',
    'Ligula RRH': 'Ligula Regional Referral Hospital'
}
```

```python
#create sentinal site column
hn['Sentinel Site'] = hn['Hospital Name'].map(hospital_map)
```

```python
#verify the changes
hn['Sentinel Site'].value_counts(dropna=False)
```

```python
#move the column to the front
column = hn.pop('Sentinel Site')

position = hn.columns.get_loc('Hospital Name')

hn.insert(position, 'Sentinel Site', column)
```
