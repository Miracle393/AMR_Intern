THERE IS RESERCHER DECISION ON 'SPECIMEN CATEGORY' INPUT, WHERE HERE I GOT TO CREATE NEW COLUMN FOR THIS ACCORDING TO [ URINE = UTI [Urinary Tract Infection] , BLOOD = BSI [Blood Stream Infection ], PUS = SSTI [Skin and Soft Tissue Infection]]

```python
sp_clean['Specimen type'].unique()
```

```python
#create dictionary first
specimen_category = {
    'urine': 'Urinary Tract Infection',
    'Blood': 'Blood Stream Infection',
    'Pus': 'Skin and Soft Tissue Infection'
}
```

```python
#create new column
sp_clean['Specimen Category'] = (
    sp_clean['Specimen type']
        .map(specimen_category)
)
```

```python
sp_clean['Specimen Category'].unique()
#sp_clean['Specimen Category']
```

```python
sp_clean.columns
```

```python
#create dictionary for specimen code
category_code = {
    'urine':'UTI',
    'Blood':'BSI',
    'Pus':'SSTI'
}
```

```python
sp_clean['Specimen Category Code'] = (
    sp_clean['Specimen type']
        .map(category_code)
)
```

```python
sp_clean['Specimen Category Code'].unique()
```

```python
# arrange the column near to specimen information group, but check first wher are they
sp_clean.columns.tolist()
```

now shifting the postion of new columns, to near other specimen information


```python
#create pop data
sc = sp_clean.pop('Specimen Category')
```

```python
#create first position
position_sc = sp_clean.columns.get_loc('Specimen type (Numeric)') + 1
```

```python
#moving the column
sp_clean.insert(position_sc, 'Specimen Category', sc)
```

```python
#check
sp_clean.columns
```

```python
#moving on to moving the specimnen category code
#create pop for value exist already
scc = sp_clean.pop('Specimen Category Code')
```

```python
#position
position_scc = sp_clean.columns.get_loc('Specimen Category') + 1
```

```python
#moving
sp_clean.insert(position_scc, 'Specimen Category Code', scc)
```
