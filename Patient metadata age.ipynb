```python
#function to standardize
def standardize_age(age):
    if pd.isna(age):
        return None
    
    age = str(age).strip().lower()
    
    # Days
    if age.endswith('d'):
        value = age.replace('d', '')
        return f"{value} Days"
    
    # Months
    elif age.endswith('m'):
        value = age.replace('m', '')
        return f"{value} Months"
    
    # Years (no letter)
    else:
        return f"{age} Years"
```

```python
pm_age['Age_standardized'] = pm_age['Age'].apply(standardize_age)
```

convert the age for easy machine language approach as 
12 Days     → 0.03 years
18 Months   → 1.5 years
31 Years    → 31 years

here workflow as;
- Keep the original raw Age column (for traceability)
- Create a standardized readable age column (Age_standardized)
- Create a numeric age column in years (Age_years) for analysis/modeling
- Create the age category based on your team's classification


```python
#function to convert to numeric
def age_to_years(age):
    if pd.isna(age):
        return None
    
    age = str(age).strip().lower()
    
    if age.endswith('d'):
        value = float(age.replace('d',''))
        return value / 365
    
    elif age.endswith('m'):
        value = float(age.replace('m',''))
        return value / 12
    
    else:
        return float(age)
```

```python
pm_age['Age_years'] = pm_age['Age'].apply(age_to_years)
```

```python
pm_age[['Age', 'Age_standardized', 'Age_years']].sample(20, random_state=42)
```

CREATING THE NEW AGE CATEGORY

```python
#fxn for age categorizing
def age_category(age):
    if pd.isna(age):
        return None
    
    if age <= 1:
        return "Infant"
    
    elif age <= 9:
        return "Children"
    
    elif age <= 19:
        return "Adolescents"
    
    elif age <= 34:
        return "Youth"
    
    elif age <= 44:
        return "Young adults"
    
    elif age <= 59:
        return "Middle age"
    
    else:
        return "Elderly"
```

```python
pm_age['Age_category'] = pm_age['Age_years'].apply(age_category)
```
